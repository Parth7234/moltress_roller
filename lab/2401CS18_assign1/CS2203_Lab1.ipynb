```python
import random

# Initialize board
board = [" " for _ in range (9) ]

def print_board () :
  print ()
  print ( board [0] , "|", board [1] , "|", board [2])
  print (" - -+ - - -+ - -")
  print ( board [3] , "|", board [4] , "|", board [5])
  print (" - -+ - - -+ - -")
  print ( board [6] , "|", board [7] , "|", board [8])
  print ()

def check_winner ( player ) :
  win_conditions = [
  (0 ,1 ,2) , (3 ,4 ,5) , (6 ,7 ,8) ,
  (0 ,3 ,6) , (1 ,4 ,7) , (2 ,5 ,8) ,
  (0 ,4 ,8) , (2 ,4 ,6)
  ]
  return any( board [ a ] == board [ b ] == board [ c ] == player
  for a , b , c in win_conditions )

def is_draw () :
  return " " not in board

def human_move () :
  while True :
    move = int( input (" Enter your move (1 -9): ") ) - 1
    if board [ move ] == " ":
      board [ move ] = "X"
      break
    else :
      print (" Invalid move , try again .")

def ai_move () :
  # Try to win
  for i in range (9) :
    if board [ i ] == " ":
      board [ i ] = "O"
      if check_winner ("O") :
        return
      board [ i ] = " "

  # Try to block human
  for i in range (9) :
    if board [ i ] == " ":
      board [ i ] = "X"
      if check_winner ("X") :
        board [ i ] = "O"
        return
      board [ i ] = " "

  # Random move
  move = random . choice ([ i for i in range (9) if board [ i ] == " "])
  board [ move ] = "O"

# Game loop
print ("Tic -Tac -Toe: You (X) vs AI (O)")

print_board ()

while True :
  human_move ()
  print_board ()
  if check_winner ("X") :
    print ("You win!")
    break
  if is_draw () :
    print ("It ’s a draw !")
    break

  ai_move ()
  print ("AI move :")
  print_board ()
  if check_winner ("O") :
    print ("AI wins !")
    break
  if is_draw () :
    print ("It ’s a draw !")
    break
```
