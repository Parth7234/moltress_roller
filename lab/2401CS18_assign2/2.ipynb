```python
import pandas as pd

def gradient(w):
    return 4 * (w - 2)**3

def gradient_descent(w, learning_rate, iterations):
    weight_history = []
    weight_history.append(round(w, 4))
    for _ in range(iterations):
        grad = gradient(w)
        w = w - learning_rate * grad
        weight_history.append(round(w, 4))
    return weight_history

w0 = -1
iterations = 5
learning_rates = [0.01, 0.1, 0.5]

results = {}

for lr in learning_rates:
    results[lr] = gradient_descent(w0, lr, iterations)

df = pd.DataFrame(results)
df.index.name = 'Iteration'
df.index += 1
print("Weight values for each learning rate after 5 iterations:")
print(df)
```
