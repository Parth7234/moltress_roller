```python
import math

x1 = 1
x2 = 1
y = 0

w11 = 0.15
w12 = 0.25
w21 = 0.20
w22 = 0.30
b1 = 0.35
b2 = 0.35

v1 = 0.40
v2 = 0.45
bo = 0.60

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

net_h1 = x1 * w11 + x2 * w21 + b1
net_h2 = x1 * w12 + x2 * w22 + b2

out_h1 = sigmoid(net_h1)
out_h2 = sigmoid(net_h2)

net_o = out_h1 * v1 + out_h2 * v2 + bo
y_hat = sigmoid(net_o)

loss = 0.5 * (y_hat - y) ** 2

print(f"Hidden neuron 1 net input: {net_h1:.4f}")
print(f"Hidden neuron 2 net input: {net_h2:.4f}")
print()
print(f"Hidden neuron 1 output: {out_h1:.4f}")
print(f"Hidden neuron 2 output: {out_h2:.4f}")
print()
print(f"Output neuron net input: {net_o:.4f}")
print(f"Final output: {y_hat:.4f}")
print()
print(f"Loss: {loss:.4f}")
```
